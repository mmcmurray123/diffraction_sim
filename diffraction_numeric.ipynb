```python
import numpy as np
import matplotlib.pyplot as plt
```

![grating_pic](params_reference.png "Grating Diagram")

```python
#Diffraction params

I_o = 100 #intensity of a single slit at the central axis
n_slits = 4 # number of slits
w_slits = 500*1e-9 # width of slits
wavelength = 630*1e-9 # wavelength of propoagting wave
l = 10 # distance to far field
x_max = 100
x = np.arange(-x_max, x_max, 0.001) # position on screen
s = 0.005*1e-6
```

```python
# the outer sinc^2 envelope
sinc_input = x*w_slits/(wavelength*l)
sinc_squared = (np.sinc(sinc_input))**2
dirichlet_input = np.pi*x*s/(wavelength * l)
dirichlet_kernel =  (np.sin(n_slits*dirichlet_input) / np.sin(dirichlet_input))**2
diffraction_pattern = I_o*sinc_squared*dirichlet_kernel
```

```python
fig, ax = plt.subplots()
ax.plot(x, diffraction_pattern)

ax.set(xlabel='position (m)', ylabel='Intensity W/m^2',
       title='Diffraction Pattern of N slits at far field')
ax.grid()

fig.savefig("diffraction_intensity.png")
plt.show()
```
