```python
import numpy as np
import matplotlib.pyplot as plt
```

To construct the intensity pattern of the diffraction grating with slits of width W and width of width G, we numerically consider the simpnler gratings, and then convolve/multiply such that we arrive at our intensity model based on widths and lengths. After doing this we arrive at an equation to plug and chug

![grating_pic](grating_picture.png "Grating Diagram")

```python
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

fig, ax = plt.subplots()
ax.plot(t, s)

ax.set(xlabel='time (s)', ylabel='voltage (mV)',
       title='About as simple as it gets, folks')
ax.grid()

fig.savefig("test.png")
plt.show()
```
